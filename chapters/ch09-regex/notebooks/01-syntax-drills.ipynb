## 1. \d — match any single digit (0-9)

**3-Sentence explanation: The characters '/d' means any single digit characters from aschii (0-9). It's used to grab every one single number from only 0-9 no letters, space, punctuations for example: re.findall(f'\d', 'Python3.12'). Because it is a re.findall, every single number it comes across will be appended to a list. ['3', '1', '2']. It's limited to only incrimints of one digit between 0-9.

```python
import re
re.findall(r'\d', 'abc123def456')
```

## 3. \s — match any whitespace character (space, tab, newline)

**3-Sentence explanation:** although \s matches each single space in a string, it also matches a new line \n and a tab \t. It is usually used to split words separated by space and outputting a list of each word or character separated by a space. Just like the others \s is limited by a single space and you would need to utilize \s+ and you must be mindful of the fact that newlines and tabs will also be extracted witch might catch you off guard if you scaning a file with multiple lines, thus is why you would use [ \t].

```python
import re
re.findall(r'\s', 'one two\tthree\nfour')
```

```python
import re
# my drill — try variations here
re.split(r'\s+', 'Hello\nWorld')
```
