## 1. \d — match any single digit (0-9)

Because it is a re.findall, every single number it comes across will be appended to a list. ['3', '1', '2']. It's limited to only incrimints of one digit between 0-9.

```python
import re
re.findall(r'\d', 'abc123def456')
```

## 2. \w — match any word character (letters, digits, underscore)

**3-Sentence explanation:** \w is a shorthand character class that matches any single word character — letters (including accented and other-alphabet letters), digits, and the underscore. You use it when you need to grab letters/numbers without writing out [A-Za-z0-9_], like pulling identifiers, usernames, or filenames out of text. Its limit is that it only matches ONE character at a time so you need \w+ for a full word, AND it includes underscore + non-ASCII letters which can match more than you intended (e.g. r'First Name: (\w+)' will catch accented letters but not hyphens, so names like Sinéad work but Jean-Paul cuts off).

```python
import re
re.findall(r'\w', 'hello, world!')
```

```python
import re
# my drill — try variations here
re.findall(r'[\w-]+', 'mother-in-law')
```

## 6. {} — quantifier count — match exactly N (or N to M) repetitions

**3-Sentence explanation:** {} is a quantifier count — placed after a character/group, {n} means exactly n repetitions, {n,m} means between n and m, and {n,} means n or more. You use it when you need precise repetition control, like \d{3} for a 3-digit area code, \d{4,6} for a zip+4 range, or \w{8,} for passwords of 8+ characters. Its limit is that the values must be non-negative integers in low-to-high order ({5,2} errors), and like every quantifier it's greedy by default so {2,5} will grab 5 if it can.

```python
import re
re.findall(r'\d{3}', 'call 555-1234 today')
```

## 7. * — match zero or more of the previous element

**3-Sentence explanation:** * is a quantifier that matches zero or more of the previous character or group, meaning the thing is optional and can repeat any number of times. You use it when something might be missing OR appear many times, like ha* matching h, ha, haaa, or \s* to consume optional whitespace between tokens. Its limit is that it can match an empty string (zero occurrences) which can produce unexpected '' results in re.findall, and it's greedy by default so .* will gobble the whole line — use *? for lazy.

```python
import re
re.findall(r'go*d', 'gd god good goood')
```

## 8. + — match one or more of the previous element

**3-Sentence explanation:** + is a quantifier that matches one or more of the previous character or group, requiring AT LEAST one occurrence to match. You use it when the thing is mandatory but can repeat, like \d+ for any number (1+ digits), \w+ for a full word, or \s+ to collapse multiple spaces into a single split point. Its limit is that the lone character without a follow-up will be dropped (unlike *), and it's greedy by default so use +? for lazy when you need the shortest match.

```python
import re
re.findall(r'go+d', 'gd god good goood')
```

## 9. ? — match zero or one of the previous element

**3-Sentence explanation:** ? is a quantifier that matches zero or one of the previous character or group, making the thing optional. You use it for optional letters or pieces, like colou?r matching both color and colour, or https?:// matching both http and https URLs. Its limit is dual-purpose confusion — when ? appears AFTER another quantifier (+?, *?, ??) it switches that quantifier from greedy to lazy instead of meaning 'optional', which trips up beginners.

```python
import re
re.findall(r'colou?r', 'color colour')
```

## 10. [^...] — negated character class — match any character NOT in the set

**3-Sentence explanation:** [^...] is a character class with a caret right after the opening bracket that matches any single character NOT in the set you define. You use it when it's easier to describe what you DON'T want than what you do — like [^0-9] for any non-digit, [^aeiou] for any consonant, or [^"]+ to grab everything inside quotes. Its limit is the single-character ceiling (still need + outside) AND the ^ only negates when it's the FIRST character inside the brackets — anywhere else (like [a^b]) it's just a literal caret.

```python
import re
re.findall(r'[^aeiou]+', 'hello world')
```

## 11. Greedy vs Lazy quantifiers — *? +? ??

**3-Sentence explanation:** By default +, *, and ? are greedy — they match as many characters as possible while still letting the overall pattern succeed; adding a ? after them (+?, *?, ??) flips them to lazy, matching as FEW as possible. You use lazy when greedy gobbles past where you want, like scraping HTML tags — r'<.+>' on '<a><b>' returns the whole '<a><b>' (greedy), while r'<.+?>' returns just '<a>' then '<b>' (lazy). Its limit is that lazy still has to match at least the minimum required (+? still needs 1, *? can be 0), and overuse of lazy can hide bugs where you actually wanted the longer span.

```python
import re
# greedy vs lazy
print(re.findall(r'".+"', 'She said "hi" then "bye"'))    # greedy
print(re.findall(r'".+?"', 'She said "hi" then "bye"'))   # lazy
```

## 12. re.search() — find the FIRST match anywhere in the string

**3-Sentence explanation:** re.search(pattern, string) scans the entire string and returns a Match object for the FIRST place the pattern matches, or None if there is no match. You use it when you only need to know IF and WHERE a pattern occurs once — like checking if an email contains '@', finding the first phone number in a log line, or validating a format. Its limit is that it returns only ONE match (not all of them) and you must call .group() on the Match object to get the actual string — calling search on a non-match gives None which crashes if you call .group() on it.

```python
import re
m = re.search(r'\d+', 'order #4471 placed at 09:23')
print(m)
print(m.group())
```

## 13. re.match() — match ONLY at the START of the string

**3-Sentence explanation:** re.match(pattern, string) tries to match the pattern at the BEGINNING of the string only, returning a Match object on success or None otherwise. You use it for validation where the input must start a certain way — like checking a string starts with a digit, a protocol prefix, or a specific keyword. Its limit is that it ONLY anchors at the start (it does NOT search the whole string), so re.match(r'\d+', 'abc123') returns None even though digits exist — for that case you want re.search() instead.

```python
import re
print(re.match(r'\d+', '123abc'))   # matches at start
print(re.match(r'\d+', 'abc123'))   # None — digits aren't at start
```

## 14. re.findall() — return ALL non-overlapping matches as a list

**3-Sentence explanation:** re.findall(pattern, string) returns a list of every non-overlapping match of the pattern in the string, or an empty list [] if nothing matches. You use it when you want to extract every occurrence — all phone numbers in a doc, all words in a sentence, all email addresses in a log — and process them as a list. Its limit is that when the pattern contains groups (), findall returns the GROUPS instead of the full match (a list of tuples if multiple groups), which surprises beginners — use re.finditer() if you need the full Match objects.

```python
import re
re.findall(r'\d+', 'phone 555-1234, ext 99, room 204')
```

## 15. re.sub() — substitute (replace) every match with a replacement string

**3-Sentence explanation:** re.sub(pattern, replacement, string) returns a NEW string with every non-overlapping match of pattern replaced by replacement, leaving the original string untouched. You use it for cleanup and transformation — censoring sensitive numbers, normalizing whitespace, swapping date formats, or stripping HTML tags. Its limit is that strings in Python are immutable so re.sub returns a NEW string (you must reassign), AND backreferences in the replacement use \1 \2 syntax — forgetting the raw-string r'' prefix on the replacement can silently break those.

```python
import re
re.sub(r'\d', 'X', 'order 4471 at 09:23')
```

## 16. re.compile() — pre-compile a pattern into a reusable Regex object

**3-Sentence explanation:** re.compile(pattern) compiles a regex string into a Regex object that has its own .search(), .findall(), .sub(), etc. methods, so you can reuse the same pattern many times without re-parsing it. You use it when the same pattern runs many times (in a loop, across many files, or as a module constant), or when you want named flags like re.IGNORECASE or re.DOTALL passed once at compile time. Its limit is that for one-off matches it adds boilerplate with no real speed benefit (the re module caches recent patterns anyway), so its real value is readability and shared flags across calls.

```python
import re
phone_re = re.compile(r'\d{3}-\d{4}')
print(phone_re.findall('call 555-1234 or 555-9999'))
print(phone_re.search('call 555-1234').group())
```

## 17. Groups () — capture parts of a match for extraction or backreference

**3-Sentence explanation:** Parentheses () create a capturing group that remembers the matched substring, accessible via Match.group(n) where group(0) is the whole match and group(1), group(2)... are the captured groups left to right. You use them to pull pieces out of a structured pattern — area code and number from a phone, name and domain from an email, or year/month/day from a date. Its limit is that with re.findall(), groups change the return value from full matches to tuples of groups (a common gotcha), and ( and ) are metacharacters so to match a LITERAL parenthesis you must escape: \( and \).

```python
import re
m = re.search(r'(\d{3})-(\d{4})', 'phone 555-1234')
print(m.group(0))   # whole match
print(m.group(1))   # area
print(m.group(2))   # number
```

## 18. Anchors — `^`, `$`, and `\b` — match POSITIONS, not characters

**3-Sentence explanation:** Anchors match positions in the string instead of consuming characters: `^` asserts start of string (or line with re.MULTILINE), `$` asserts end of string/line, and `\b` asserts a word boundary between a word char (`\w`) and a non-word char. You use them to lock matches to specific positions — `^Error` to find lines starting with Error, `\.txt$` for filenames ending in .txt, or `\bcat\b` to match the word cat but not 'caterpillar' or 'concatenate'. Its limit is they match zero-width positions (don't consume characters), and `^`/`$` behavior changes with re.MULTILINE — without it, `^` only matches the very start of the whole string, not each line.

```python
import re
print(re.findall(r'^\w+', 'hello world'))           # ['hello']
print(re.findall(r'\w+$', 'hello world'))           # ['world']
print(re.findall(r'\bcat\b', 'cat caterpillar concatenate cat'))  # ['cat', 'cat']
```
