## 6. Normalize phone numbers to (XXX) XXX-XXXX

Take messy 10-digit US phone numbers and reformat them all to `(XXX) XXX-XXXX`.

**Limit / what breaks it:** Assumes US 10-digit numbers; ignores extensions, country codes, and non-US formats.

```python
raw = ["(415) 555-2671", "415-555-2671", "415.555.2671", "4155552671", "+1 415 555 2671"]
```

```python
import re
# TODO: write your regex here
def normalize(s):
    return s  # replace
for r in raw:
    print(r, "->", normalize(r))
```

```python
import re
def normalize(s):
    digits = re.sub(r"\D", "", s)
    if len(digits) == 11 and digits.startswith("1"):
        digits = digits[1:]
    if len(digits) != 10:
        return None
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"
for r in raw:
    print(r, "->", normalize(r))
```

```python
import re
pattern = (
    r"(?P<ip>\S+) \S+ (?P<user>\S+) \[(?P<time>[^\]]+)\] "
    r'"(?P<method>\S+) (?P<path>\S+) (?P<proto>[^"]+)" '
    r'(?P<status>\d+) (?P<size>\S+) "(?P<referer>[^"]*)" "(?P<agent>[^"]*)"'
)
m = re.match(pattern, line)
print(m.groupdict() if m else None)
```
